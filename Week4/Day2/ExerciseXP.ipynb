Exercise 1 : Matrix Operations
Instructions
In this exercise, you’ll work with a 3x3 matrix. Here’s a brief explanation of the concepts:

Determinant: The determinant is a value that can be computed from the elements of a square matrix. It provides important information about the matrix, such as whether it has an inverse, and is used in various areas like linear algebra and calculus. To understand more about it you can watch this video.
Inverse of a Matrix: The inverse of a matrix is a matrix that, when multiplied with the original matrix, results in an identity matrix. Not all matrices have inverses. The inverse is crucial in solving systems of linear equations.
Create a 3x3 matrix and perform the following operations:

Calculate the determinant.
Find the inverse of the matrix.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
 
# Create a 3x3 matrix
matrix = [
    [1, 2, 3],
    [0, 1, 4],
    [5, 6, 0]
]
 
print("Original Matrix:")
for row in matrix:
    print(row)
print()

matrix_np = np.array(matrix)

# Calculate with NumPy
determinant = np.linalg.det(matrix_np)
determinant_rounded = round(determinant, 10)  # ← ROUND HERE
print(f"NumPy determinant: {determinant}")
print(f"NumPy rounded: {determinant_rounded}")
print()

# Rule of Sarrus
a, b, c = matrix[0]
d, e, f = matrix[1]
g, h, i = matrix[2]

# Apply formula
determinant_sarrus = (a*e*i + b*f*g + c*d*h) - (c*e*g + a*f*h + b*d*i)
print(f"Rule of Sarrus: {determinant_sarrus}")

# Find the inverse
inverse = np.linalg.inv(matrix_np)
print("Inverse Matrix:")
print(inverse)
```

Exercise 6 : Basic Hypothesis Testing
Instructions
Create a sample dataset to test the effectiveness of a new training program on employee productivity:


import numpy as np

# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.


Given a dataset, formulate a simple hypothesis and test it using basic statistical functions in NumPy.

```python
# Productivity scores of employees before the training program
productivity_before = np.random.normal(loc=50, scale=10, size=30)

# Productivity scores of the same employees after the training program
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Your task is to formulate a hypothesis regarding the training program's effectiveness 
# and test it using basic statistical functions in NumPy.
#########################################################################################################################################################################


print("My hypothesis is that the training did help and we can calculate how much it helped improve by percentage.")


print(productivity_before)
print(productivity_after)

# #1: t-statistic
differences = productivity_after - productivity_before
mean_diff = np.mean(differences)
std_diff = np.std(differences, ddof=1)
se = std_diff / np.sqrt(len(differences))
t_stat = mean_diff / se

# #2: Mean + Percentage
mean_before = np.mean(productivity_before)
mean_after = np.mean(productivity_after)
diff = mean_after - mean_before
percent = (diff / mean_before) * 100

print(f"t-statistic: {t_stat:.2f}")
print(f"Improvement: {diff:.2f} points ({percent:.2f}%)")

if t_stat > 3:
    print("HYPOTHESIS CONFIRMED! Training program is EFFECTIVE!")
```

Exercise 8 : Time Series Data Manipulation
Instructions
Generate time series data for the year 2023. Demonstrate slicing for the following intervals:

January to March
April to June
July to September
October to December
Generate a time series data for a specific period and demonstrate how to slice this data for different intervals.



```python
# Create date range for all of 2023
dates = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')  # D = Daily

print(dates)
print(f"Total dates: {len(dates)}")

# Create random values for each date
values = np.random.randint(90, 150, size=len(dates))

print(values[:10]) 

time_series = pd.Series(values, index=dates)

print(time_series)

q1 = time_series['2023-01':'2023-03']  # January to March
q2 = time_series['2023-04':'2023-06']  # April to June
q3 = time_series['2023-07':'2023-09']  # July to September
q4 = time_series['2023-10':'2023-12']  # October to December

# Analyze each quarter
print(f"Q1: {len(q1)} days, Mean: {q1.mean():.2f}")
print(f"Q2: {len(q2)} days, Mean: {q2.mean():.2f}")
print(f"Q3: {len(q3)} days, Mean: {q3.mean():.2f}")
print(f"Q4: {len(q4)} days, Mean: {q4.mean():.2f}")
```
