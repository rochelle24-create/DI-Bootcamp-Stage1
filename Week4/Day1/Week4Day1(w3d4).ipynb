## Numpy Library

A list of numbers in python is very similar to a 1D numpy array

```python
l = [1,2,3,4,5]
```

2D Array ->

```python
li = [
    [5.2, 3.0, 4.5],
    [9.1, 0.1, 0.3]
]
print(li)
```

3D Array ->

```python
threeD_array = [
    [
        [1,4,7],
        [2,9,7],
        [1,3,0],
        [9,6,9]
    ],
    [
        [2,3,4],
        [1,1,5],
        [1,1,2],
        [1,1,3]
    ]
]
```

```python
import numpy as np
```

We can make arrays in a bunch of ways, the simplest is using python lists

```python
ar1 = np.array(l)
print(ar1)
print(l)
```

```python
ar2 = np.array(li)
print(ar2)
print('----------------')
print(li)
```

We can analyze the shape and size of our arrays

```python
print(ar2.shape, ar2.size)
```

Python lists don't quite understand the dimensions

```python
print(len(li))
```

```python
print(ar2.dtype)
```

## Accessing Items in Arrays

```python
ar2
```

```python
ar2[0,2]
```

```python
threeD_array = np.array(threeD_array)
print(threeD_array)
```

```python
threeD_array[1,1,2]
```

```python
threeD_array[1,1,:] # equivalent to threeD_array[1,1,0:3]
```

We can get that whole horizontal slice by taking all depth layers 

```python
threeD_array[:,1,:]
```

We can also access items in an array by giving a list of indexes

```python
ar1[np.array((1,3))]
```

In a one dimensional array we can take every other item by using the step option

```python
ar1
```

```python
ar1[0:5:2] # ar1[start:stop:step] here step is 2
```

We can take every other row/floor by using the step option with our indexing/slicing

```python
threeD_array[:,::2,:]
```

We can access items with conditions

```python
ar1 > 2
```

```python
ar1[ar1 > 2]
```

```python
threeD_array == 1
```

```python
threeD_array[threeD_array == 1]
```

We can change values in the array very simply

```python
ar1[3] = 42
ar1
```

```python
threeD_array[threeD_array == 1] = 0
threeD_array
```

```python
ar1[1:4] = -10
ar1
```

```python
ar1[1:4] = [5,4,3]
ar1
```

Converting an image to grayscale with PIL and mean

```python
from PIL import Image
image = Image.open('Soldier.png')
```

```python
imagear = np.array(image)
imagear.shape
```

```python
imagear[0,:10,:10]
```

```python
gray = imagear.mean(axis=2)
gray = gray.round()
gray = gray.astype(np.uint8)
grayimage = Image.fromarray(gray)
grayimage.save('graysoldier.png')
```
