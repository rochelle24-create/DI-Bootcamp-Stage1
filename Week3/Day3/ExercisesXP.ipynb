Exercise 2: Descriptive Statistics
Task: Given a sample dataset, calculate the mean, median, variance, and standard deviation using SciPy.
Sample Dataset:
data = [12, 15, 13, 12, 18, 20, 22, 21]

```python
import numpy as np
from scipy import stats

# Sample Dataset
data = [12, 15, 13, 12, 18, 20, 22, 21]

print("Dataset:", data)
print()

# Calculate statistics
mean = np.mean(data)
median = np.median(data)
variance = np.var(data, ddof=1)
std_dev = np.std(data, ddof=1)

print(f"Mean:      {mean:.4f}")
print(f"Median:    {median:.4f}")
print(f"Variance:  {variance:.4f}")
print(f"Std Dev:   {std_dev:.4f}")
```

Exercise 3: Understanding Distributions
Task: Generate a normal distribution using SciPy with a mean of 50 and a standard deviation of 10. Plot this distribution.
Code Example:
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# complete your code here

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Generate normal distribution
x = np.linspace(10, 90, 100)
y = norm.pdf(x, loc=50, scale=10)

# Plot
plt.plot(x, y)
plt.fill_between(x, y, alpha=0.3)
plt.title('Normal Distribution (μ=50, σ=10)')
plt.xlabel('Value')
plt.ylabel('Probability')
plt.show()
```

Exercise 4: T-Test Application
Task: Perform a T-test on two sets of randomly generated data.
Code Example:
data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

# complete your code here

```python
import numpy as np
from scipy.stats import ttest_ind

data1 = np.random.normal(50, 10, 100)
data2 = np.random.normal(60, 10, 100)

t_stat, p_value = ttest_ind(data1, data2)

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.6f}")
print(f"Significant: {p_value < 0.05}")
```

Exercise 5: Linear Regression Analysis
Objective: Apply linear regression to a dataset and interpret the results.

Task: Given a dataset of housing prices (house_prices) and their corresponding sizes (house_sizes), use linear regression to predict the price of a house given its size.
Dataset:
house_sizes: [50, 70, 80, 100, 120] (in square meters)
house_prices: [150,000, 200,000, 210,000, 250,000, 280,000] (in currency units)
Questions:
What is the slope and intercept of the regression line?
Predict the price of a house that is 90 square meters.
Interpret the meaning of the slope in the context of housing prices.

```python
import numpy as np
from scipy.stats import linregress

house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

slope, intercept, r_value, p_value, std_err = linregress(house_sizes, house_prices)

print(f"Slope: {slope:.2f}")
print(f"Intercept: {intercept:.2f}")
print(f"R-squared: {r_value**2:.4f}")
print()

# Predict price for 90 square meters
predicted_price = slope * 90 + intercept
print(f"Predicted price for 90 sq meters: {predicted_price:,.2f}")
print()

# Interpretation
print(f"Interpretation:")
print(f"- For every 1 sq meter increase, price increases by {slope:,.2f} units")
print(f"- Base price (intercept): {intercept:,.2f}")
```

Exercise 6: Understanding ANOVA
Objective: Test understanding of ANOVA and its application.

Task: Three different fertilizers are applied to three separate groups of plants to test their effectiveness. The growth in centimeters is recorded.
Dataset:
fertilizer_1: [5, 6, 7, 6, 5]
fertilizer_2: [7, 8, 7, 9, 8]
fertilizer_3: [4, 5, 4, 3, 4]
Questions:
Perform an ANOVA test on the given data. What are the F-value and P-value?
Based on the P-value, do the fertilizers have significantly different effects on plant growth?
Explain what would happen if the P-value were greater than 0.05.

```python
import numpy as np
from scipy.stats import f_oneway

fertilizer_1 = np.array([5, 6, 7, 6, 5])
fertilizer_2 = np.array([7, 8, 7, 9, 8])
fertilizer_3 = np.array([4, 5, 4, 3, 4])

f_value, p_value = f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"F-value: {f_value:.4f}")
print(f"P-value: {p_value:.6f}")
print()

if p_value < 0.05:
    print("Result: Fertilizers have SIGNIFICANTLY different effects (p < 0.05)")
else:
    print("Result: Fertilizers have NO significant difference (p >= 0.05)")
print()

print("Interpretation:")
print("- P < 0.05: Reject null hypothesis - fertilizers are different")
print("- P >= 0.05: Fail to reject - no significant difference between fertilizers")
```

Exercise 7: Probability Distributions (Optional)
Work with a binomial distribution: calculate probabilities for different numbers of successes.
Example: Calculating the probability of getting exactly 5 heads in 10 coin flips.


```python
import numpy as np
from scipy.stats import binom

# Parameters: n=10 flips, p=0.5 probability of heads
n = 10
p = 0.5

# Probability of exactly 5 heads
prob_5 = binom.pmf(5, n, p)
print(f"P(exactly 5 heads in 10 flips): {prob_5:.4f}")
print()

# Probability for different numbers of heads
print("Probability distribution:")
for k in range(0, 11):
    prob = binom.pmf(k, n, p)
    print(f"P(X = {k:2d}): {prob:.4f}")
print()

# Probability of at most 5 heads
prob_at_most_5 = binom.cdf(5, n, p)
print(f"P(X <= 5): {prob_at_most_5:.4f}")

# Probability of more than 5 heads
prob_more_than_5 = 1 - binom.cdf(5, n, p)
print(f"P(X > 5): {prob_more_than_5:.4f}")
```

Calculate the Pearson and Spearman correlation coefficients between two variables in a dataset.
Example in code: data = pd.DataFrame({'age': [23, 25, 30, 35, 40], 'income': [35000, 40000, 50000, 60000, 70000]})


```python
import numpy as np
from scipy.stats import pearsonr, spearmanr

age = np.array([23, 25, 30, 35, 40])
income = np.array([35000, 40000, 50000, 60000, 70000])

# Pearson correlation
pearson_r, pearson_p = pearsonr(age, income)
print(f"Pearson correlation: {pearson_r:.4f}")
print(f"P-value: {pearson_p:.6f}")
print()

# Spearman correlation
spearman_r, spearman_p = spearmanr(age, income)
print(f"Spearman correlation: {spearman_r:.4f}")
print(f"P-value: {spearman_p:.6f}")
print()

print("Interpretation:")
print("- Correlation close to 1: Strong positive relationship")
print("- Correlation close to 0: No relationship")
print("- Correlation close to -1: Strong negative relationship")
```
