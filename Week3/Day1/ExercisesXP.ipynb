```python
# Exercise 9 : Export a dataframe to excel format and JSON format.
# Create a simple dataframe.
# Export the dataframe to an excel file.
# Export the dataframe to a JSON file.

import pandas as pd
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 28, 32],
    'Department': ['Sales', 'IT', 'HR', 'Finance', 'Marketing'],
    'Salary': [50000, 60000, 55000, 65000, 52000],
    'Experience': [2, 5, 8, 3, 6]
}
df = pd.DataFrame(data)

df.to_excel('exported_data.xlsx', index=False, sheet_name='Employee Data')

df.to_json('exported_data.json', orient='records', indent=2)
```
